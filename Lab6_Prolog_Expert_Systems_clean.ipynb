```python
class PrologSimulator:
    def __init__(self):
        self.facts = set()
        self.rules = []
    def add_fact(self, fact):
        self.facts.add(fact)
    def add_rule(self, head, body):
        self.rules.append((head, body))
    def query(self, goal):
        if goal in self.facts:
            return True
        for head, body in self.rules:
            if head == goal:
                if all(self.query(condition) for condition in body):
                    return True
        return False
print("Prolog Simulator ready!")
```

```python
class MedicalExpertSystem:
    def __init__(self):
        self.symptoms = {}
        self.diseases = {
            'flu': ['fever', 'cough', 'headache'],
            'measles': ['fever', 'rash', 'sore_throat'],
            'covid': ['fever', 'cough', 'shortness_of_breath'],
            'heart_issue': ['chest_pain', 'shortness_of_breath']
        }
        self.treatments = {
            'flu': 'Rest, fluids, and fever reducers',
            'measles': 'Rest, fluids, and vitamin A supplements',
            'covid': 'Isolation, rest, and medical monitoring',
            'heart_issue': 'Immediate medical attention required'
        }
    def add_symptom(self, patient, symptom):
        if patient not in self.symptoms:
            self.symptoms[patient] = []
        self.symptoms[patient].append(symptom)
    def diagnose(self, patient):
        if patient not in self.symptoms:
            return None
        patient_symptoms = set(self.symptoms[patient])
        for disease, required_symptoms in self.diseases.items():
            if all(symptom in patient_symptoms for symptom in required_symptoms):
                return disease
        return 'unknown'
    def recommend_treatment(self, disease):
        return self.treatments.get(disease, 'Consult a medical professional')
mes = MedicalExpertSystem()
mes.add_symptom('john', 'fever')
mes.add_symptom('john', 'cough')
mes.add_symptom('john', 'headache')
mes.add_symptom('mary', 'fever')
mes.add_symptom('mary', 'rash')
mes.add_symptom('mary', 'sore_throat')
mes.add_symptom('bob', 'chest_pain')
mes.add_symptom('bob', 'shortness_of_breath')
print("\nDiagnosis Results:")
print("="*50)
for patient in ['john', 'mary', 'bob']:
    diagnosis = mes.diagnose(patient)
    treatment = mes.recommend_treatment(diagnosis)
    print(f"\nPatient: {patient.title()}")
    print(f"Symptoms: {', '.join(mes.symptoms[patient])}")
    print(f"Diagnosis: {diagnosis}")
    print(f"Treatment: {treatment}")
```

```python
class InteractiveExpertSystem:
    def __init__(self):
        self.mes = MedicalExpertSystem()
    def consultation(self, patient_name):
        print(f"\n{'='*50}")
        print(f"Medical Consultation for {patient_name}")
        print(f"{'='*50}\n")
        symptoms_list = [
            'fever', 'cough', 'headache', 'rash', 
            'sore_throat', 'chest_pain', 'shortness_of_breath'
        ]
        print("Available symptoms:")
        for i, symptom in enumerate(symptoms_list, 1):
            print(f"{i}. {symptom}")
        example_symptoms = ['fever', 'cough', 'headache']
        print(f"\nPatient symptoms: {', '.join(example_symptoms)}")
        for symptom in example_symptoms:
            self.mes.add_symptom(patient_name, symptom)
        diagnosis = self.mes.diagnose(patient_name)
        treatment = self.mes.recommend_treatment(diagnosis)
        print(f"\nDiagnosis: {diagnosis.upper()}")
        print(f"Recommended Treatment: {treatment}")
        print(f"\n{'='*50}\n")
ies = InteractiveExpertSystem()
ies.consultation('Alice')
```

```python
class SimpleGrammarChecker:
    def __init__(self):
        self.determiners = {'the', 'a', 'an'}
        self.nouns = {'cat', 'dog', 'ball', 'mouse', 'bird'}
        self.verbs = {'chases', 'sees', 'catches', 'eats'}
    def is_determiner(self, word):
        return word.lower() in self.determiners
    def is_noun(self, word):
        return word.lower() in self.nouns
    def is_verb(self, word):
        return word.lower() in self.verbs
    def parse_noun_phrase(self, words, index):
        start = index
        if index < len(words) and self.is_determiner(words[index]):
            index += 1
        if index < len(words) and self.is_noun(words[index]):
            index += 1
            return index
        return -1
    def parse_verb_phrase(self, words, index):
        if index >= len(words) or not self.is_verb(words[index]):
            return -1
        index += 1
        np_end = self.parse_noun_phrase(words, index)
        if np_end != -1:
            return np_end
        return index
    def parse_sentence(self, sentence):
        words = sentence.lower().split()
        index = self.parse_noun_phrase(words, 0)
        if index == -1:
            return False, "Invalid noun phrase"
        end = self.parse_verb_phrase(words, index)
        if end == -1:
            return False, "Invalid verb phrase"
        if end != len(words):
            return False, "Extra words after complete sentence"
        return True, "Valid sentence"
checker = SimpleGrammarChecker()
test_sentences = [
    "the cat chases the dog",
    "cat sees mouse",
    "a dog catches ball",
    "the ball chases",  
    "cat dog",  
    "chases the cat",  
]
print("\nGrammar Checking Results:")
print("="*50)
for sentence in test_sentences:
    valid, message = checker.parse_sentence(sentence)
    status = "✓" if valid else "✗"
    print(f"{status} '{sentence}' - {message}")
```

```python
class SemanticAnalyzer:
    def __init__(self):
        self.animals = {'cat', 'dog', 'mouse', 'bird'}
        self.objects = {'ball', 'stick', 'toy'}
        self.can_eat = {
            'cat': {'mouse', 'bird', 'food'},
            'dog': {'food'},
            'bird': {'seeds', 'worms'}
        }
    def is_animal(self, word):
        return word in self.animals
    def is_object(self, word):
        return word in self.objects
    def can_chase(self, chaser, chased):
        return (self.is_animal(chaser) and 
                (self.is_animal(chased) or self.is_object(chased)) and
                chaser != chased)
    def can_perform_eat(self, eater, food):
        return eater in self.can_eat and food in self.can_eat[eater]
    def validate_action(self, subject, verb, object_):
        if verb == 'chases':
            return self.can_chase(subject, object_)
        elif verb == 'eats':
            return self.can_perform_eat(subject, object_)
        elif verb == 'sees':
            return True  
        return False
analyzer = SemanticAnalyzer()
test_actions = [
    ('cat', 'chases', 'mouse'),
    ('dog', 'chases', 'ball'),
    ('ball', 'chases', 'cat'),  
    ('cat', 'eats', 'mouse'),
    ('cat', 'eats', 'dog'),  
    ('dog', 'sees', 'bird'),
]
print("\nSemantic Validation:")
print("="*50)
for subject, verb, obj in test_actions:
    valid = analyzer.validate_action(subject, verb, obj)
    status = "✓" if valid else "✗"
    print(f"{status} '{subject} {verb} {obj}' - {'Valid' if valid else 'Semantically invalid'}")
```

```python
class QuestionAnsweringSystem:
    def __init__(self):
        self.is_a = {
            'cat': 'mammal',
            'dog': 'mammal',
            'bird': 'animal',
            'mammal': 'animal'
        }
        self.has = {
            'cat': {'fur', 'whiskers', 'tail'},
            'dog': {'fur', 'tail'},
            'bird': {'feathers', 'wings'},
            'mammal': {'backbone'},
            'animal': {'cells'}
        }
        self.can = {
            'cat': {'meow', 'purr'},
            'dog': {'bark', 'wag_tail'},
            'bird': {'fly', 'chirp'}
        }
    def get_ancestors(self, entity):
        ancestors = []
        current = entity
        while current in self.is_a:
            parent = self.is_a[current]
            ancestors.append(parent)
            current = parent
        return ancestors
    def has_property(self, entity, property_):
        if entity in self.has and property_ in self.has[entity]:
            return True
        for ancestor in self.get_ancestors(entity):
            if ancestor in self.has and property_ in self.has[ancestor]:
                return True
        return False
    def can_do(self, entity, ability):
        return entity in self.can and ability in self.can[entity]
    def answer_question(self, question_type, entity, attribute):
        if question_type == 'has':
            return self.has_property(entity, attribute)
        elif question_type == 'can':
            return self.can_do(entity, attribute)
        elif question_type == 'is':
            return attribute in self.get_ancestors(entity) or entity == attribute
        return False
qa_system = QuestionAnsweringSystem()
questions = [
    ('has', 'cat', 'fur', 'Does a cat have fur?'),
    ('has', 'cat', 'backbone', 'Does a cat have a backbone?'),  
    ('has', 'bird', 'fur', 'Does a bird have fur?'),
    ('can', 'dog', 'bark', 'Can a dog bark?'),
    ('can', 'cat', 'fly', 'Can a cat fly?'),
    ('is', 'cat', 'animal', 'Is a cat an animal?'),  
    ('is', 'cat', 'bird', 'Is a cat a bird?'),
]
print("\nQuestion Answering:")
print("="*50)
for q_type, entity, attr, question in questions:
    answer = qa_system.answer_question(q_type, entity, attr)
    print(f"Q: {question}")
    print(f"A: {'Yes' if answer else 'No'}\n")
```
