```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Image, display
print(f"OpenCV Version: {cv2.__version__}")
def display_image(img, title="Image", cmap=None):
    plt.figure(figsize=(10, 8))
    if cmap:
        plt.imshow(img, cmap=cmap)
    elif len(img.shape) == 3:
        plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    else:
        plt.imshow(img, cmap='gray')
    plt.title(title)
    plt.axis('off')
    plt.tight_layout()
    plt.show()
def display_multiple(images, titles, cmap=None):
    n = len(images)
    fig, axes = plt.subplots(1, n, figsize=(5*n, 5))
    if n == 1:
        axes = [axes]
    for i, (img, title) in enumerate(zip(images, titles)):
        if cmap:
            axes[i].imshow(img, cmap=cmap)
        elif len(img.shape) == 3:
            axes[i].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
        else:
            axes[i].imshow(img, cmap='gray')
        axes[i].set_title(title)
        axes[i].axis('off')
    plt.tight_layout()
    plt.show()
```

```python
img = cv2.imread('sample_image.jpg', cv2.IMREAD_GRAYSCALE)
_, thresh1 = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)
_, thresh2 = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY_INV)
_, thresh3 = cv2.threshold(img, 127, 255, cv2.THRESH_TRUNC)
_, thresh4 = cv2.threshold(img, 127, 255, cv2.THRESH_TOZERO)
adaptive_mean = cv2.adaptiveThreshold(img, 255, cv2.ADAPTIVE_THRESH_MEAN_C,
                                      cv2.THRESH_BINARY, 11, 2)
adaptive_gaussian = cv2.adaptiveThreshold(img, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
                                         cv2.THRESH_BINARY, 11, 2)
_, otsu = cv2.threshold(img, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
fig, axes = plt.subplots(2, 4, figsize=(16, 8))
axes = axes.ravel()
images = [img, thresh1, thresh2, thresh3, thresh4, adaptive_mean, adaptive_gaussian, otsu]
titles = ['Original', 'Binary', 'Binary Inv', 'Truncate', 'To Zero', 
          'Adaptive Mean', 'Adaptive Gaussian', "Otsu's"]
for i, (image, title) in enumerate(zip(images, titles)):
    axes[i].imshow(image, cmap='gray')
    axes[i].set_title(title)
    axes[i].axis('off')
plt.tight_layout()
plt.show()
```

```python
img = np.zeros((200, 200), dtype=np.uint8)
cv2.rectangle(img, (50, 50), (150, 150), 255, -1)
noise = np.random.randint(0, 2, (200, 200), dtype=np.uint8) * 255
img = cv2.bitwise_or(img, noise)
kernel = np.ones((5, 5), np.uint8)
erosion = cv2.erode(img, kernel, iterations=1)
dilation = cv2.dilate(img, kernel, iterations=1)
opening = cv2.morphologyEx(img, cv2.MORPH_OPEN, kernel)
closing = cv2.morphologyEx(img, cv2.MORPH_CLOSE, kernel)
gradient = cv2.morphologyEx(img, cv2.MORPH_GRADIENT, kernel)
tophat = cv2.morphologyEx(img, cv2.MORPH_TOPHAT, kernel)
blackhat = cv2.morphologyEx(img, cv2.MORPH_BLACKHAT, kernel)
fig, axes = plt.subplots(2, 4, figsize=(16, 8))
axes = axes.ravel()
images = [img, erosion, dilation, opening, closing, gradient, tophat, blackhat]
titles = ['Original', 'Erosion', 'Dilation', 'Opening', 'Closing', 'Gradient', 'Top Hat', 'Black Hat']
for i, (image, title) in enumerate(zip(images, titles)):
    axes[i].imshow(image, cmap='gray')
    axes[i].set_title(title)
    axes[i].axis('off')
plt.tight_layout()
plt.show()
```

```python
img = np.zeros((400, 600, 3), dtype=np.uint8)
img[:] = (255, 255, 255)
cv2.rectangle(img, (50, 50), (150, 150), (0, 0, 0), -1)
cv2.circle(img, (300, 100), 50, (0, 0, 0), -1)
pts = np.array([[450, 50], [500, 150], [400, 150]], np.int32)
cv2.fillPoly(img, [pts], (0, 0, 0))
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
contours, hierarchy = cv2.findContours(cv2.bitwise_not(gray), cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
contour_img = img.copy()
cv2.drawContours(contour_img, contours, -1, (0, 255, 0), 3)
rect_img = img.copy()
for contour in contours:
    x, y, w, h = cv2.boundingRect(contour)
    cv2.rectangle(rect_img, (x, y), (x+w, y+h), (255, 0, 0), 2)
print(f"Number of contours found: {len(contours)}")
for i, contour in enumerate(contours):
    area = cv2.contourArea(contour)
    perimeter = cv2.arcLength(contour, True)
    print(f"Contour {i+1}: Area={area:.2f}, Perimeter={perimeter:.2f}")
display_multiple([img, contour_img, rect_img], ['Original', 'Contours', 'Bounding Boxes'])
```

```python
img = np.zeros((400, 600, 3), dtype=np.uint8)
img[:] = (200, 200, 200)
cv2.rectangle(img, (150, 100), (250, 200), (0, 255, 0), -1)
cv2.circle(img, (400, 150), 40, (255, 0, 0), -1)
template = img[120:180, 170:230].copy()
result = cv2.matchTemplate(img, template, cv2.TM_CCOEFF_NORMED)
min_val, max_val, min_loc, max_loc = cv2.minMaxLoc(result)
h, w = template.shape[:2]
top_left = max_loc
bottom_right = (top_left[0] + w, top_left[1] + h)
matched_img = img.copy()
cv2.rectangle(matched_img, top_left, bottom_right, (0, 0, 255), 3)
display_multiple([img, template, matched_img], ['Original', 'Template', 'Matched'])
print(f"Match confidence: {max_val:.4f}")
```

```python
img = np.ones((400, 400, 3), dtype=np.uint8) * 255
cv2.circle(img, (200, 200), 100, (200, 200, 200), -1)  
cv2.circle(img, (170, 180), 15, (0, 0, 0), -1)  
cv2.circle(img, (230, 180), 15, (0, 0, 0), -1)  
cv2.ellipse(img, (200, 240), (40, 20), 0, 0, 180, (0, 0, 0), 2)  
try:
    face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    faces = face_cascade.detectMultiScale(gray, 1.1, 4)
    for (x, y, w, h) in faces:
        cv2.rectangle(img, (x, y), (x+w, y+h), (0, 255, 0), 2)
    print(f"Faces detected: {len(faces)}")
except Exception as e:
    print(f"Haar Cascade not available: {e}")
display_image(img, "Face Detection Demo")
```

```python
img = np.zeros((400, 600, 3), dtype=np.uint8)
img[:] = (255, 255, 255)
pts = np.array([[150, 100], [450, 100], [500, 300], [100, 300]], np.int32)
cv2.fillPoly(img, [pts], (100, 150, 200))
src_points = np.float32([[150, 100], [450, 100], [500, 300], [100, 300]])
dst_points = np.float32([[150, 100], [450, 100], [450, 300], [150, 300]])
matrix = cv2.getPerspectiveTransform(src_points, dst_points)
result = cv2.warpPerspective(img, matrix, (600, 400))
display_multiple([img, result], ['Original (Trapezoid)', 'Transformed (Rectangle)'])
```

```python
img1 = np.zeros((300, 300, 3), dtype=np.uint8)
cv2.rectangle(img1, (50, 50), (250, 250), (255, 255, 255), -1)
img2 = np.zeros((300, 300, 3), dtype=np.uint8)
cv2.circle(img2, (150, 150), 100, (255, 255, 255), -1)
bitwise_and = cv2.bitwise_and(img1, img2)
bitwise_or = cv2.bitwise_or(img1, img2)
bitwise_xor = cv2.bitwise_xor(img1, img2)
bitwise_not = cv2.bitwise_not(img1)
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.ravel()
images = [img1, img2, bitwise_and, bitwise_or, bitwise_xor, bitwise_not]
titles = ['Image 1', 'Image 2', 'AND', 'OR', 'XOR', 'NOT (Image 1)']
for i, (image, title) in enumerate(zip(images, titles)):
    axes[i].imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    axes[i].set_title(title)
    axes[i].axis('off')
plt.tight_layout()
plt.show()
```
